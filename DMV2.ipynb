```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
```

```python
# ============================================
# 4. Handle Missing Values
# ============================================

# Numerical columns → median
numeric_cols = df.select_dtypes(include=np.number).columns

for col in numeric_cols:
    df[col] = df[col].fillna(df[col].median())


# Categorical columns → mode
categorical_cols = df.select_dtypes(include="object").columns

for col in categorical_cols:
    if df[col].isnull().sum() > 0:
        df[col] = df[col].fillna(df[col].mode()[0])


print("\nMissing values after handling:")
print(df.isnull().sum())
```

```python
# Remove leading/trailing spaces
categorical_cols = df.select_dtypes(include="object").columns

for col in categorical_cols:
    df[col] = df[col].str.strip()


# Standardize text values
for col in categorical_cols:
    df[col] = df[col].str.lower()


print("\nCategorical values standardized.")
```

```python

# ============================================
# 7. Convert Data Types
# ============================================

# Convert Customer ID to string
df["Customer ID"] = df["Customer ID"].astype(str)

# Convert Zip Code to string
df["Zip Code"] = df["Zip Code"].astype(str)

# Convert numeric columns
numeric_columns = [
    "Age",
    "Number of Dependents",
    "Number of Referrals",
    "Tenure in Months",
    "Avg Monthly Long Distance Charges",
    "Avg Monthly GB Download",
    "Monthly Charge",
    "Total Charges",
    "Total Refunds",
    "Total Extra Data Charges",
    "Total Long Distance Charges",
    "Total Revenue"
]

for col in numeric_columns:
    df[col] = pd.to_numeric(df[col], errors="coerce")


print("\nData types after conversion:")
print(df.dtypes)
```

```python
# ============================================
# 8. Handle Outliers using IQR
# ============================================

numeric_cols = df.select_dtypes(include=np.number).columns

for col in numeric_cols:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    # Cap extreme values
    df[col] = df[col].clip(lower_limit, upper_limit)


print("\nOutliers handled using IQR method.")
```

```python
# ============================================
# 9. Feature Engineering
# ============================================

# Average revenue per month
df["Revenue_per_Month"] = (
    df["Total Revenue"] /
    df["Tenure in Months"].replace(0, 1)
)

# Total additional charges
df["Total_Extra_Charges"] = (
    df["Total Refunds"] +
    df["Total Extra Data Charges"] +
    df["Total Long Distance Charges"]
)

# Create age group
df["Age_Group"] = pd.cut(
    df["Age"],
    bins=[0, 25, 40, 60, 100],
    labels=["Young", "Adult", "Middle_Aged", "Senior"]
)

# Create tenure group
df["Tenure_Group"] = pd.cut(
    df["Tenure in Months"],
    bins=[-1, 12, 24, 48, 72],
    labels=["New", "Medium", "Long", "Very_Long"]
)

print("\nNew features created:")

print(
    df[
        [
            "Revenue_per_Month",
            "Total_Extra_Charges",
            "Age_Group",
            "Tenure_Group"
        ]
    ].head()
)
```

```python
# ============================================
# 10. Encode Categorical Variables
# ============================================

# Remove Customer ID because it is only an identifier
df_model = df.drop(columns=["Customer ID"])


# One-hot encoding
df_encoded = pd.get_dummies(
    df_model,
    drop_first=True,
    dtype=int
)

print("\nEncoded dataset shape:")
print(df_encoded.shape)
```

```python
# ============================================
# 11. Separate Features and Target
# ============================================

# Customer Status contains the churn information
print("\nCustomer Status values:")
print(df["Customer Status"].value_counts())


# Create binary churn target
df_encoded["Churn"] = (
    df["Customer Status"]
    .str.lower()
    .map({
        "churned": 1,
        "stayed": 0
    })
)

status_columns = [
    col for col in df_encoded.columns
    if col.startswith("Customer Status_")
]

df_encoded = df_encoded.drop(columns=status_columns)


# Separate X and y
X = df_encoded.drop(columns=["Churn"])
y = df_encoded["Churn"]
```

```python
# ============================================
# 12. Scale Numerical Features
# ============================================

scaler = StandardScaler()

X_scaled = pd.DataFrame(
    scaler.fit_transform(X),
    columns=X.columns
)


# ============================================
# 13. Train-Test Split
# ============================================

X_train, X_test, y_train, y_test = train_test_split(
    X_scaled,
    y,
    test_size=0.20,
    random_state=42
)

print("\nTraining data:", X_train.shape)
print("Testing data:", X_test.shape)
```

```python

# ============================================
# 14. Export Cleaned Dataset
# ============================================

df.to_csv(
    "Telecom_Customer_Churn_Cleaned.csv",
    index=False
)

print("\nCleaned dataset exported successfully!")

# Export encoded dataset
df_encoded.to_csv(
    "Telecom_Customer_Churn_Encoded.csv",
    index=False
)

print("Encoded dataset exported successfully!")
```

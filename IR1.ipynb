```python
import nltk
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
from nltk.tokenize import word_tokenize

nltk.download('stopwords')
nltk.download('punkt')
nltk.download('punkt_tab') # Added to resolve LookupError
```

```python
"""### Stop Word Removal

Stop words are common words (like 'the', 'is', 'a') that often carry little significant meaning and can be removed to reduce noise in text processing. This function tokenizes the text and removes these words.
"""

def remove_stopwords(text):
    stop_words = set(stopwords.words('english'))
    word_tokens = word_tokenize(text)
    filtered_text = [w for w in word_tokens if not w.lower() in stop_words]
    return " ".join(filtered_text)
```

```python
"""### Stemming

Stemming is the process of reducing inflected (or sometimes derived) words to their word stem, base or root form—generally a written word form. The stem is not necessarily identical to the morphological root of the word; it is usually an affix of a word.
"""

def apply_stemming(text):
    ps = PorterStemmer()
    word_tokens = word_tokenize(text)
    stemmed_text = [ps.stem(w) for w in word_tokens]
    return " ".join(stemmed_text)
```

```python
"""### Demonstration of Preprocessing Functions"""

example_text = "This is an example sentence, showing off the stop word removal and stemming functionalities of NLTK. Running and learning new things are great!"

print(f"Original Text: {example_text}\n")
```

```python
# Stop word removal
text_without_stopwords = remove_stopwords(example_text)
print(f"After Stop Word Removal: {text_without_stopwords}\n")
```

```python
# Stemming
stemmed_text = apply_stemming(text_without_stopwords)
print(f"After Stemming (on text without stopwords): {stemmed_text}")
```
